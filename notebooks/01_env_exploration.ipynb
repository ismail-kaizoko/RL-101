```python
from env.drone_env import DroneEnv2D
import numpy as np
```

```python
env = DroneEnv2D(world_size=20, curriculum_level=0, render_mode="human")
```

```python
obs, info = env.reset(seed=42)
print("Observation shape:", obs.shape)   # should be (14,) — 6 + 8 rays
print("Action space:", env.action_space)
```

```python
total_reward = 0
for step in range(200):
    action = env.action_space.sample()   # random policy
    obs, reward, terminated, truncated, info = env.step(action)
    total_reward += reward
    env.render()
    if terminated or truncated:
        print(f"Episode ended at step {step}, total reward: {total_reward:.2f}")
        break

env.close()
```
