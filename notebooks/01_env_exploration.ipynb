```python
import sys, importlib
from pathlib import Path

ROOT = Path.cwd().parent
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from env.drone_env import DroneEnv2D
import numpy as np
```

```python
env = DroneEnv2D(world_size=20, level=2, render_mode="human", fps=8)
```

```python
obs, info = env.reset(seed=42)
print("Observation shape:", obs.shape)   # should be (14,) — 6 + 8 rays
print("Action space:", env.action_space)
```

```python
import pygame

obs, _ = env.reset(seed=42)
total_reward = 0.0
env.render()   # warm up pygame before the event loop

running = True
for step in range(500):
    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            running = False
            break
    if not running:
        break

    action = env.action_space.sample()
    obs, reward, terminated, truncated, info = env.step(action)
    total_reward += reward
    env.render()

    if terminated or truncated:
        print(f"Episode ended at step {step+1}  |  total reward: {total_reward:.2f}")
        break

env.close()
```
